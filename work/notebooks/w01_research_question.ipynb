## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The question: Which existing pages should be reviewed first for refresh, expansion, or pruning?
Unit of analysis: a single page.
The decision: which pages a content editor tackles first this week, out of potentially thousands of pages — since nobody has time to review everything.
The action: an editor gets a ranked list of pages with a suggested action per page (refresh, expand, prune, monitor) and a reason code explaining why it was flagged (e.g. "declining traffic with steady demand," "stale but still getting impressions").
The cost of a wrong call: two ways this can go wrong. False positive — an editor spends hours refreshing a page that wouldn't have improved anyway, wasting time. False negative — a page that's quietly losing traffic gets ignored and keeps declining until it's much harder to recover. Since editor time is limited, the real cost isn't just "wrong page," it's "right page not reviewed because the wrong one got picked instead."
Why data/ML helps at all: with a small number of pages, a person could just eyeball them. At the scale FlyRank operates on (tens of thousands of pages across clients), nobody can manually check each one — a model that ranks pages by "worth checking first" turns an impossible manual task into a short, reviewable list.
